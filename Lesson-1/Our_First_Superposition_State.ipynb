```python
#Neste experimento vamos testar as portas lógicas quânticas nos circuitos para ver o resultados que obtemos.
#Nessa primeira parte vamos utilizar a porta identidade que não muda nada no qubit.
```

```python
from qiskit import QuantumRegister, QuantumCircuit
from qiskit_aer import StatevectorSimulator, QasmSimulator
from biblioteca import Wavefunction

S_simulator = StatevectorSimulator()

q_1 = QuantumRegister(1)
circuit_1 = QuantumCircuit(q_1)

circuit_1.id(q_1[0])

display(circuit_1.draw(output='mpl'))
job_1 = S_simulator.run(circuit_1)
result_1 = job_1.result()
result_1.get_statevector()
```

```python
#Nesta segunda parte vamos utilizar a porta X (NOT gate) no qubit, essa porta é equivalente  a porta NOT da lógica clássica então ela inverte o valor do qubit.
```

```python
q_2 = QuantumRegister(1)
circuit_2 = QuantumCircuit(q_2)

circuit_2.x(q_2[0])

display(circuit_2.draw(output='mpl'))
Wavefunction(circuit_2) #utilizando função da biblioteca para imprimir o estado do vertor com ket.
```

```python
#neste terceiro experimento vamos criar nosso primeiro estado de superposição. Em especial vamos utilizar a porta Hadamard que separa um qubit em 50-50 probabilidade de distribuição entre os estados
#|0> e |1>.
```

```python
q_3 = QuantumRegister(1)
circuit_3 = QuantumCircuit(q_3)

circuit_3.h(q_3[0])

display(circuit_3.draw(output='mpl'))
Wavefunction(circuit_3)
```
